```python
def light_on(pin_number:int, time_on:int):
    '''
    Activates a pin
    
    Inputs:
        Pin number
    
    Outputs:
        Specific pin number activated
    '''
    
    GPIO.setmode(GPIO.BCM)
    
    PIN = pin_number
    
    GPIO.setup(PIN, GPIO.OUT)
    
    print("Pin 17 is on ")
    
    GPIO.output(PIN, GPIO.HIGH)
    
    time.sleep(time_on)
    
    GPIO.output(PIN, GPIO.LOW)
    
    GPIO.cleanup()
```

```python
from logging import raiseExceptions
def set_light_color(color_code:int):
  """
  Convert color number code into a color.

  Placeholder function for a later version that will execute an actual color change in a circuit

  Parameters:
  color_code : int
    This is a code (0,1,2) that will translate to a color

  Returns:
  color_switch : str
    This will be the color that the function returns. Valid values:
    0 -> Green
    1 -> yellow
    2 -> Red
  """
  light_dict = {
      0 : "Green",
      1 : "Yellow",
      2 : "Red"
  }

  valid_lights_list = list(light_dict.keys())

  if color_code in valid_lights_list:
    pass
  else:
    raise ValueError(
        f"{color_code} is not a valid value. Valid values are {list(light_dict)}."
    )

  color_switch = light_dict[color_code]
  #print(color_switch)



  return color_switch

```

```python
def stop_light(color_code:int):
  """
  Take a color and return what needs to happen on the circuit next

  Parameters:
  color : str
    color that is given

  Returns:
  -------
  next_action : dict
    A dictionary that has the wait time and the next color
  """

  color_lookup_dict = {
      0 : {"wait_time": 10, "next_color":1},
      1 : {"wait_time": 2, "next_color":2},
      2 : {"wait_time": 15, "next_color":0}
  }
   

  valid_lookup_keys_list = list(color_lookup_dict.keys())

  next_action = color_lookup_dict[color_code]

  return next_action
```

```python
def run_stop_light(starting_color:int = 0):
  """
  Runs the stop light

  Parameters:
  -----------
  starting_color : int, default = 0
    the color that you want the light to start at

  """

  run_iterations = 0
  while run_iterations < 10:

    if run_iterations == 0:
      current_color = starting_color

    # First, we want to set the color to the starting color
    set_light_color(current_color)
    # Returning the wait time and the next color
    next_action_dict = stop_light(current_color)
    
    if current_color == 0:
        light_on(17, 10)
        print("Green")
        #time.sleep(time_on)
        current_color = next_action_dict["next_color"]
        next_action_dict = stop_light(current_color)
    
    elif current_color == 1:
        light_on(27, 2)
        print("Yellow")
        #time.sleep(time_on)
        current_color = next_action_dict["next_color"]
        next_action_dict = stop_light(current_color)
        
    
    elif current_color == 2:
        light_on(22, 15)
        print("Red")
        #time.sleep(time_on)
        current_color = next_action_dict["next_color"]
        next_action_dict = stop_light(current_color)
        

    run_iterations += 1
```
