```python
import RPi.GPIO as GPIO
import time
```

```python
def pir_input():
    
    GPIO.setmode(GPIO.BCM)
    
    while True:
        
        start_time = time.time()
        
        GPIO.setup(27, GPIO.IN)
        
        if GPIO.input(27) == 1:
            print("Motion")
            end_time = time.time()
            total_time = end_time - start_time
            print(total_time)
            break
        else:
            pass
        

    
```

```python
pir_input()
```
