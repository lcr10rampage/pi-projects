```python
import RPi.GPIO as GPIO
import time
```

```python
def light_on(pin_number:int):
    '''
    Activates a pin
    
    Inputs:
        Pin number
    
    Outputs:
        Specific pin number activated
    '''
    
    GPIO.setmode(GPIO.BCM)
    
    PIN = pin_number
    
    GPIO.setup(PIN, GPIO.OUT)
    
    print("Pin 17 is on ")
    
    GPIO.output(PIN, GPIO.HIGH)
    
    time.sleep(5)
    
    GPIO.output(PIN, GPIO.LOW)
    
    GPIO.cleanup()
```

```python
light_on(17)
```
