```python
def keypad_logic(code:str):
    """
    Runs four times. Each time, it checks if the user inputed the right number in on the keypad. If they get all four right, a green light lights up. If not, a red light lights up.
    
    Return:
    ---------
        A Green light lit up if right, a red light if not right.
    """
    code_list = list(code)
    user_list = []
    run_iterations = 0
    
    while run_iterations < 4:
        # Ask the user to type a number
        user_number = input("Type a number that you think it is. ")

        #Check and see if that number is in the correct spot of code_list
        if user_number == code_list[run_iterations]:
            print(user_number)
            user_list.append(user_number)
            print(code_list)
            print(user_list)
        else:
            print("Red")
            break
        
        run_iterations += 1
        
        if user_list == code_list:
            print("Green")
            break
        else:
            pass
            
```

```python
def check_key_pressed():
    columns = [17,27,22,5]
    rows = [24,23,25]
    GPIO.setmode(GPIO.BCM)
    while True:
        
        for ROW in rows:
            GPIO.setup(ROW, GPIO.OUT)
            GPIO.output(ROW, GPIO.HIGH)
            
            for COL in columns:
                GPIO.setup(COL, GPIO.IN, pull_up_down=GPIO.PUD_DOWN)
                if GPIO.input(COL):
                    GPIO.output(ROW, GPIO.LOW)
                    return True
            
    GPIO.cleanup()
    return False
```

```python
def keypad(code:str):
    
    GPIO.cleanup()
    
    Pins = [17,27,22,5,23,24,25]
    
    keypad_number_dict = {
        0 : {"number_to_activate" : 24, "number_to_input" : 5},
        1 : {"number_to_activate" : 23, "number_to_input" : 17},
        2 : {"number_to_activate" : 24, "number_to_input" : 17},
        3 : {"number_to_activate" : 25, "number_to_input" : 17},
        4 : {"number_to_activate" : 23, "number_to_input" : 27},
        5 : {"number_to_activate" : 24, "number_to_input" : 27},
        6 : {"number_to_activate" : 25, "number_to_input" : 27},
        7 : {"number_to_activate" : 23, "number_to_input" : 22},
        8 : {"number_to_activate" : 24, "number_to_input" : 22},
        9 : {"number_to_activate" : 25, "number_to_input" : 22}
    }
    
    code_list = list(code)
    user_list = []
    run_iterations = 0
    GPIO.setmode(GPIO.BCM)
    
    while run_iterations < 4:
        # Activate the pin of the first number
        # Ask the user to type a number
        print("Type a number")
        
        activate_number = code_list[run_iterations]
        dict_number = int(activate_number)
        
        current_active_pin = keypad_number_dict[dict_number]
        activate_pin = current_active_pin["number_to_activate"]
        
        GPIO.setup(activate_pin, GPIO.OUT)
        GPIO.output(activate_pin, GPIO.HIGH)
        
        current_input_pin = keypad_number_dict[dict_number]
        input_pin = current_input_pin["number_to_input"]
        
        GPIO.setup(input_pin, GPIO.IN, pull_up_down=GPIO.PUD_DOWN)
        
        
        #Check and see if that number is in the correct spot of code_list
        while True:
            
            if check_key_pressed() == True:
            

                if GPIO.input(input_pin):
                    print(activate_number)
                    print("Blue")
                    user_list.append(activate_number)
                    GPIO.output(activate_pin, GPIO.LOW)
                    break

                else:
                    print("Red")
                    return False
            
                        
        run_iterations += 1
        time.sleep(.5)
        
        if user_list == code_list:
            print("Green")
            break
    GPIO.cleanup()    
```
