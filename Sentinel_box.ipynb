```python
def keypad(code:str):
    
    GPIO.cleanup()
    
    Pins = [17,27,22,5,23,24,25]
    
    keypad_number_dict = {
        0 : {"number_to_activate" : 24, "number_to_input" : 5},
        1 : {"number_to_activate" : 24, "number_to_input" : 17},
        2 : {"number_to_activate" : 23, "number_to_input" : 17},
        3 : {"number_to_activate" : 25, "number_to_input" : 17},
        4 : {"number_to_activate" : 24, "number_to_input" : 22},
        5 : {"number_to_activate" : 24, "number_to_input" : 27},
        6 : {"number_to_activate" : 25, "number_to_input" : 27},
        7 : {"number_to_activate" : 23, "number_to_input" : 22},
        8 : {"number_to_activate" : 23, "number_to_input" : 27},
        9 : {"number_to_activate" : 25, "number_to_input" : 22}
    }
    
    code_list = list(code)
    user_list = []
    run_iterations = 0
    GPIO.setmode(GPIO.BCM)
    
    while run_iterations < 4:
        # Activate the pin of the first number
        # Ask the user to type a number
        print("Type a number")
        
        activate_number = code_list[run_iterations]
        dict_number = int(activate_number)
        
        current_active_pin = keypad_number_dict[dict_number]
        activate_pin = current_active_pin["number_to_activate"]
        
        GPIO.setup(activate_pin, GPIO.OUT)
        GPIO.output(activate_pin, GPIO.HIGH)
        time.sleep(.001)
        
        current_input_pin = keypad_number_dict[dict_number]
        input_pin = current_input_pin["number_to_input"]
        
        GPIO.setup(input_pin, GPIO.IN, pull_up_down=GPIO.PUD_DOWN)
        
        
        #Check and see if that number is in the correct spot of code_list
        while True:
            
            if check_key_pressed() == True:
            

                if GPIO.input(input_pin):
                    time.sleep(.05)
                    if GPIO.input(input_pin):
                        user_list.append(activate_number)
                        GPIO.output(activate_pin, GPIO.LOW)
                        GPIO.cleanup(input_pin)
                        time.sleep(.5)
                        print(str(user_list))
                        break

                    else:
                        return False
            
                        
        run_iterations += 1
        time.sleep(.5)
        
        if user_list == code_list:
            print("Green")
            break
    GPIO.cleanup()
```

```python
def pulse():
    GPIO.setmode(GPIO.BCM)
    
    GPIO.setup(27, GPIO.OUT)
    
    GPIO.setup(17, GPIO.IN)
    
    

    
    while True:
        
        GPIO.output(27, GPIO.HIGH)
        time.sleep(0.00001)
        GPIO.output(27, GPIO.LOW)

        while GPIO.input(17) == 0:
            pass
        #print("on")
        start_time =time.time()
        

        while GPIO.input(17) == 1:
            pass
        #print("off")
        end_time = time.time()
        
        total_time = end_time - start_time
        distance_in_cm = (total_time * 34300) / 2
        #print(f"Total distance: {distance_in_cm} cm")
        return distance_in_cm
        
        break


        time.sleep(2)
    GPIO.cleanup()
```

```python
def keypad_interaction(code):
    
    lcd = CharLCD(
    i2c_expander='PCF8574',
    address=0x27,
    port=1,
    cols=16,
    rows=2

    )
    
    while True:
        
        lcd.write_string("Type Your 4     Digit Passcode")

        keypad_entry = int(input("Type your password"))        
        
        if keypad_entry == code:
            lcd.clear()
            lcd.write_string("Correct Code.   Opening")
            time.sleep(3)
            lcd.clear()
            break
        elif keypad_entry != code:
            lcd.clear()
            lcd.write_string("Incorrect Code. Try Again")
            time.sleep(2)
            lcd.clear()
            
```
