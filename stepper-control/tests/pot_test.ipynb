# Pot reading test (ADS7830)

**Wiring**
| ADS7830 pin | Connect to |
|---|---|
| VCC | 3.3V (Pi pin 1) |
| GND | GND (Pi pin 6) |
| SDA | Pi pin 3 (GPIO 2) |
| SCL | Pi pin 5 (GPIO 3) |
| A0 | pot middle leg |

Pot outer legs go to 3.3V and GND.

Run the cell below and turn the pot. Press the stop (■) button to end it.
A working chip shows `raw` moving between 0 and 255 as you turn.

```python
import time
from smbus2 import SMBus

ADS7830_ADDR = 0x48   # default address of the ADS7830 module

def read_ads7830(bus, channel=0):
    """Return 0..255 for the given channel (single-ended, internal reference on)."""
    cmd = 0x84 | ((((channel << 2) | (channel >> 1)) & 0x07) << 4)
    return bus.read_byte_data(ADS7830_ADDR, cmd)

def show_pot(channel=0, interval=1.0):
    with SMBus(1) as bus:
        print(f"Reading ADS7830 A{channel}. Turn the pot. Press stop to end.")
        try:
            while True:
                raw = read_ads7830(bus, channel)
                print(f"raw={raw:3d}   value={raw/255:.3f}   ~{raw/255*3.3:.2f} V")
                time.sleep(interval)
        except KeyboardInterrupt:
            print("Stopped.")

show_pot()
```
